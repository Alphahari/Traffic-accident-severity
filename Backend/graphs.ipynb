```python
import os
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Load the data
df = pd.read_csv('cleaned_dataset.csv')  # Replace with your actual file path

# Define the directory where you want to save the graphs
save_dir = './graphs'  # Replace with your desired directory path

# Create the directory if it doesn't exist
if not os.path.exists(save_dir):
    os.makedirs(save_dir)

# Save the second plot: Severity by Day of the Week
plt.figure(figsize=(8, 5))
sns.countplot(x='Day_of_Week', hue='Accident_Severity', data=df, palette='Set1')
plt.title('Accident Severity by Day of the Week')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.savefig(os.path.join(save_dir, 'severity_by_day_of_week.png'))  # Correct file name
plt.show()
plt.close()

# Save the third plot: Severity by Road Class
plt.figure(figsize=(8, 5))
sns.countplot(x='1st_Road_Class', hue='Accident_Severity', data=df, palette='coolwarm')
plt.title('Severity by Road Class')
plt.ylabel('Count')
plt.savefig(os.path.join(save_dir, 'severity_by_road_class.png'))  # Correct file name
plt.show()
plt.close()

# Save Light Conditions vs Severity
plt.figure(figsize=(10, 6))
sns.countplot(y='Light_Conditions', hue='Accident_Severity', data=df, palette='viridis')
plt.title('Light Conditions vs Severity')
plt.tight_layout()
plt.savefig(os.path.join(save_dir, 'light_conditions_vs_severity.png'))  # Correct file name
plt.show()
plt.close()

# Save Weather Conditions vs Severity
plt.figure(figsize=(10, 6))
sns.countplot(y='Weather_Conditions', hue='Accident_Severity', data=df, palette='husl')
plt.title('Severity by Weather Conditions')
plt.xlabel('Count')
plt.tight_layout()
plt.savefig(os.path.join(save_dir, 'severity_by_weather_conditions.png'))  # Correct file name
plt.show()
plt.close()

# Save Road Type vs Severity
plt.figure(figsize=(8, 5))
sns.countplot(x='Road_Type', hue='Accident_Severity', data=df, palette='Accent')
plt.title('Road Type vs Severity')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(os.path.join(save_dir, 'road_type_vs_severity.png'))  # Correct file name
plt.show()
plt.close()

# Save Speed Limit vs Severity (Boxplot)
plt.figure(figsize=(8, 5))
sns.boxplot(x='Accident_Severity', y='Speed_limit', data=df, palette='Set3')
plt.title('Speed Limit vs Accident Severity')
plt.tight_layout()
plt.savefig(os.path.join(save_dir, 'speed_limit_vs_severity.png'))  # Correct file name
plt.show()
plt.close()

# Save Junction Detail vs Severity
plt.figure(figsize=(10, 6))
sns.countplot(y='Junction_Detail', hue='Accident_Severity', data=df, palette='cubehelix')
plt.title('Junction Detail vs Severity')
plt.tight_layout()
plt.savefig(os.path.join(save_dir, 'junction_detail_vs_severity.png'))  # Correct file name
plt.show()
plt.close()
```

```python
from sklearn.preprocessing import LabelEncoder

encoded_df = df.copy()
for col in encoded_df.columns:
    if encoded_df[col].dtype == 'object':
        encoded_df[col] = LabelEncoder().fit_transform(encoded_df[col])

plt.figure(figsize=(12, 10))
sns.heatmap(encoded_df.corr(), annot=True, fmt=".2f", cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.savefig(os.path.join(save_dir, 'Correlation Heatmap'))
plt.show()
plt.close()
```
